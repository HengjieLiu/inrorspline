Four independent GaussianDIR endpoints plus identity and saved SITReg, evaluated on the same pair. Set `GAUSSIANDIR_REVIEW_ROOT` to the external generated report directory. This source notebook stays output-free.

```python
import os, json, csv, html
from pathlib import Path
from IPython.display import display, Markdown, Image, HTML
root = Path(os.environ['GAUSSIANDIR_REVIEW_ROOT']).resolve()
assert root.is_relative_to(Path(os.environ['ARTIFACT_ROOT']).resolve())
report = (root / 'report.md').read_text()
display(Markdown(report.split('![Accuracy and regularity]')[0]))
```

```python
display(Image(filename=str(root / 'accuracy_regularity.png')))
display(Image(filename=str(root / 'convergence.png')))
display(Image(filename=str(root / 'slices.png')))
```

```python
def show_table(headers, rows):
    heading = '<tr>' + ''.join('<th>' + html.escape(str(v)) + '</th>' for v in headers) + '</tr>'
    body = ''.join('<tr>' + ''.join('<td>' + html.escape(str(v)) + '</td>' for v in row) + '</tr>' for row in rows)
    display(HTML('<div style="overflow:auto;max-height:900px"><table>' + heading + body + '</table></div>'))
with (root / 'all_metrics.csv').open() as stream:
    metrics = list(csv.DictReader(stream))
display(Markdown('## Complete scalar metric comparison'))
show_table(['metric'] + [r['method'] for r in metrics], [[key] + [r[key] for r in metrics] for key in metrics[0] if key != 'method'])
display(Markdown('## All per-label metrics and eligibility'))
with (root / 'per_label.csv').open() as stream:
    labels = list(csv.DictReader(stream))
show_table(list(labels[0]), [list(row.values()) for row in labels])
```
