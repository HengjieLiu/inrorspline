Full-box supervised direct-DVF fitting, seven families, six budgets, two saved targets. Each independent fit has at most20 minutes of active work. Open the external executed notebook for results.
Filled markers indicate convergence; hollow markers indicate unresolved fits. Missing points and incomplete model growth are explicitly excluded from completed-budget curves.

```python
import os, json, csv, html
from pathlib import Path
from IPython.display import display, Image, Markdown, HTML
def show_csv(path):
    with open(path) as stream: rows = list(csv.reader(stream))
    content = ''.join('<tr>' + ''.join('<td>' + html.escape(value) + '</td>' for value in row) + '</tr>' for row in rows)
    display(HTML('<div style="max-height:500px;overflow:auto"><table>' + content + '</table></div>'))
root = Path(globals().get('RUN_ROOT') or os.environ['P82_RUN_ROOT'])
out = root / 'reports'
results = json.loads((out / 'results.json').read_text())
display(Markdown('**Coverage:** ' + str(results['counts'])))
display(Markdown('**Original cps2 reconstruction control:** ' + str(results['native_cps2_replay']['full_rmse_mm']) + ' mm'))
show_csv(out / 'results.csv')
```

```python
for name in ['full_box_curves', 'foreground_curves', 'paired_versions', 'error_maps_rank000_sitreg', 'error_maps_rank000_unbounded']:
    display(Markdown('### ' + name.replace('_', ' ')))
    display(Image(filename=str(out / (name + '.png'))))
show_csv(out / 'stage_history.csv')
```
