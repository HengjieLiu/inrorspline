# Motion Pattern Reference Review

This notebook displays the generated coronal motion-pattern panels and manifest summary.


```python
from pathlib import Path
import json
from IPython.display import Image, display, Markdown

ROOT = Path.cwd()
manifest = json.loads((ROOT / 'motion_pattern_manifest.json').read_text())
display(Markdown(f"Generated cases: `{len(manifest['cases'])}`"))
display(Image(filename=str(ROOT / 'motion_pattern_coronal_summary.png')))
display(Image(filename=str(ROOT / 'motion_pattern_coronal_reference.png')))
display(Image(filename=str(ROOT / 'motion_pattern_coronal_vector_reference.png')))
```

```python
for item in manifest['per_case_pngs']:
    display(Markdown(f"## {item['case_key']}"))
    display(Image(filename=str(ROOT / item['path'])))
for item in manifest['per_case_vector_pngs']:
    display(Markdown(f"## {item['case_key']} vector field"))
    display(Image(filename=str(ROOT / item['path'])))
```

```python
for case in manifest['cases']:
    display(Markdown(
        f"**{case['display_label']}**: slice axis `{case['slice_axis_zyx']}`, "
        f"slice index `{case['slice_index']}`, field `{case['source_field']}`.  "
        f"Orientation: {case['display_orientation']}  "
        f"Backward grid: `{case['backward_grid_mode']}`.  "
        f"Forward grid: `{case['forward_grid_mode']}`; {case['forward_grid_component_mapping']}"
    ))
```
